```python

#Practicing Variables

# Declaring variables
name = "Yusra"
age = 23
marks = 87.5
is_student = True

# Displaying variable values
print("Student Information")
print("Name:", name)
print("Age:", age)
print("Marks:", marks)
print("Is Student:", is_student)
```

In this program, I created four variables to store student information.
- name = "Yusra" stores the student's name. Since it contains text, its data type is string.
- age = 22 stores the student's age as an integer.
- marks = 87.5 stores the student's marks as a float because it contains a decimal value.
- is_student = True stores a Boolean value to show whether the person is a student or not.
After declaring these variables, I used the print() function to display their values on the screen.

The program displays the student information in an organized way.

What I learned: I learned that variables are used to store information so that we can easily access and use that information later in a program. I also understood that Python allows us to store different types of values in different variables.

```python

# Data Types

name = "Yusra"
age = 22
height = 5.4
is_student = True

# Displaying values and their data types
print("Name:", name)
print("Data Type:", type(name))

print("\nAge:", age)
print("Data Type:", type(age))

print("\nHeight:", height)
print("Data Type:", type(height))

print("\nIs Student:", is_student)
print("Data Type:", type(is_student))
```

In this program, I created four variables with different data types.
- name = "Yusra" represents a string (str), which is used to store text.
- age = 22 represents an integer (int), which is used to store whole numbers.
- height = 5.4 represents a float (float), which is used to store decimal numbers.
- is_student = True represents a Boolean (bool), which stores either True or False.
To display the values, I used the print() function.

I also used the Python type() function to identify the data type of each variable. For example, type(name) tells us that the name variable is a string.

The \n used inside the print statements adds a new line before displaying the next information. It makes the output easier to read.

```python

#Operators

a = 10
b = 5

# Arithmetic Operators
print("Arithmetic Operators")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)

# Comparison Operators
print("\nComparison Operators")
print("a is greater than b:", a > b)
print("a is equal to b:", a == b)
print("a is less than b:", a < b)

# Logical Operators
print("\nLogical Operators")
print("AND:", a > 5 and b < 10)
print("OR:", a < 5 or b < 10)
print("NOT:", not(a > b))
```

In this program, I assigned two numerical values to variables a and b.I then demonstrated three types of operators.

1. Arithmetic Operators

These operators are used to perform mathematical calculations.
- a + b performs addition and gives 15.
- a - b performs subtraction and gives 5.
- a * b performs multiplication and gives 50.
- a / b performs division and gives 2.0.
- a % b calculates the remainder after division and gives 0.
2. Comparison Operators

These operators compare two values and return either True or False.
- a > b checks whether a is greater than b. The result is True.
- a == b checks whether both values are equal. The result is False.
- a < b checks whether a is less than b. The result is False.
3. Logical Operators

These operators are used to combine or reverse conditions.
- and returns True when both conditions are true.
- or returns True when at least one condition is true.
- not reverses the result of a condition. If the condition is True, it returns False.

For example, a > 5 and b < 10 returns True because both conditions are satisfied.

```python

# Conditional Statements

marks = float(input("Enter your marks (0-100): "))

# Checking conditions
if marks >= 80 and marks <= 100:
    print("Grade: A")
    print("Excellent Performance!")

elif marks >= 70:
    print("Grade: B")
    print("Good Performance!")

elif marks >= 60:
    print("Grade: C")
    print("Average Performance!")

elif marks >= 50:
    print("Grade: D")
    print("You have passed.")

elif marks >= 0:
    print("Grade: F")
    print("You have failed.")

else:
    print("Invalid marks! Please enter a value between 0 and 100.")
```

This asks the user to enter their marks. The float() function converts the input into a decimal number.

After taking the input, I used if, elif, and else statements to check different conditions.
- if marks >= 80 and marks <= 100: checks whether the marks are between 80 and 100. If the condition is true, the program displays Grade A.
- elif marks >= 70: checks whether the marks are 70 or above if the first condition was false. It assigns Grade B.
- elif marks >= 60: assigns Grade C.
- elif marks >= 50: assigns Grade D and displays that the student has passed.
- elif marks >= 0: assigns Grade F and displays that the student has failed.
- else: handles negative marks by displaying an invalid marks message.
The program checks conditions from top to bottom. Once a condition becomes true, it executes that block and skips the remaining conditions.

```python


print("===== STUDENT RESULT SYSTEM =====")

# Variables and Input
name = input("Enter student name: ")
age = int(input("Enter student age: "))

# Data Types
print("\nStudent Information")
print("Name:", name)
print("Age:", age)
print("Name Data Type:", type(name))
print("Age Data Type:", type(age))

# Operators
marks1 = float(input("\nEnter marks for Subject 1: "))
marks2 = float(input("Enter marks for Subject 2: "))

total = marks1 + marks2
average = total / 2

print("\nResult Calculation")
print("Total Marks:", total)
print("Average Marks:", average)

# Conditional Statements
if average >= 80 and average <= 100:
    print("Grade: A")
    print("Result: Excellent")

elif average >= 70:
    print("Grade: B")
    print("Result: Good")

elif average >= 60:
    print("Grade: C")
    print("Result: Average")

elif average >= 50:
    print("Grade: D")
    print("Result: Pass")

elif average >= 0:
    print("Grade: F")
    print("Result: Fail")

else:
    print("Invalid marks entered!")
```

In this program, I combined variables, data types, user input, operators, and conditional statements to create a simple student result system.




```python

#  Online Shopping Cart

customer_name = "Sara"
product_name = "Laptop"
price = 125000.50
quantity = 2

# List of purchased products
products = ["Laptop", "Mouse", "Keyboard"]

total_price = price * quantity

print("ONLINE SHOPPING CART")
print("Customer:", customer_name)
print("Product:", product_name)
print("Price:", price)
print("Quantity:", quantity)
print("Total Price:", total_price)
print("Products:", products)

print("\nProduct List Data Type:", type(products))
```

In this program, I created a simple online shopping cart to store customer and product information.
- customer_name = "Sara" stores the customer's name as a string.
- product_name = "Laptop" stores the name of the selected product.
- price = 125000.50 stores the laptop's price as a float because it contains a decimal value.
- quantity = 2 stores the number of laptops purchased as an integer.
- products = ["Laptop", "Mouse", "Keyboard"] creates a list containing different product names. A list allows us to store multiple values in one variable.
- total_price = price * quantity calculates the total cost by multiplying the price by the quantity.
After that, I used the print() function to display all the shopping details, including the customer name, product name, price, quantity, and total price.

At the end, I used type(products) to check the data type of the products variable. It displays list, which confirms that the variable contains a Python list
